## 1. Lists

A list stores an ordered collection of items. Lists are **mutable**, which means that you can change, add, or remove items after creating the list.

Lists use square brackets `[]`. Each item has an index:
- The first item has index `0`.
- Negative indexing counts from the end: `-1` is the last item.

```python
products = ["Laptop", "Keyboard", "Mouse", "Monitor", "Printer"]

print(products)
print(products[0])
print(products[2])
print(products[4])

print(products[-1])
print(products[-2])
```

### Updating and managing list items

- `append(item)` adds an item to the end.
- `insert(index, item)` adds an item at a specified position.
- `remove(item)` removes the first matching item.
- `len(list)` returns the number of items.
- Slicing with `[start:stop]` returns a portion of the list. The stop index is excluded.

Note: `remove()` changes the existing list and returns `None`. Therefore, call it on its own line rather than passing it to `print()`.

```python
products = ["Laptop", "Keyboard", "Mouse", "Monitor"]

products.append("Printer")
products.insert(1, "USB Cable")
products.remove("Printer")

print(products)
print("Number of products:", len(products))
print("First two products:", products[:2])
```

## 2. Tuples

A tuple stores an ordered collection of items, but it is **immutable**: its items cannot be reassigned after the tuple is created.

Tuples use parentheses `()`. They are useful when a group of values should remain fixed, such as coordinates, fixed configuration values, or a record returned by a database query.

```python
sales_record = ("INV001", "Laptop", 5, 12500.00)

print(sales_record)
print(sales_record[0])
print(sales_record[1])

# This assignment would raise a TypeError because tuples are immutable:
# sales_record[0] = "INV002"
```

## 3. Dictionaries

A dictionary stores information as **key-value pairs**. Dictionaries use curly braces `{}`.

In the example, `"employee_id"` is a key and `101` is its value. A dictionary is useful for representing a business record with named attributes.

```python
employee = {
    "employee_id": 101,
    "name": "Thabo",
    "department": "Data Analytics",
    "salary": 18500.50,
    "is_permanent": True
}

print(employee)
```

### Accessing dictionary values

You can access a value using its key in square brackets, for example `employee["name"]`.

You can also use `.get()`. One advantage of `.get()` is that it can return a default value when a key does not exist, instead of raising a `KeyError`.

```python
print(employee["name"])
print(employee["salary"])

print(employee.get("department"))
print(employee.get("location", "Not specified"))
```

### Modifying dictionaries

Assigning a new value to an existing key updates that value. Assigning a value to a new key adds a new key-value pair.

Dictionary methods:
- `.keys()` returns the keys.
- `.values()` returns the values.
- `.items()` returns key-value pairs.

```python
employee["salary"] = 20000.00
employee["location"] = "Johannesburg"

print(employee)
print(employee.keys())
print(employee.values())
print(employee.items())
```

### Why dictionaries matter in data work

APIs and JSON files commonly represent records using key-value structures. The example below resembles the kind of customer record that could be received from an external system.

```python
customer = {
    "customer_id": 1001,
    "customer_name": "John",
    "city": "Johannesburg",
    "total_orders": 12
}

print(customer)
```

## 4. Sets

A set stores **unique values**. If duplicate values are supplied, only one instance of each value is retained. Sets are unordered, so do not rely on a particular display order or position.

Sets are useful for basic data-quality checks, such as identifying unique categories or city names.

```python
cities = {"Johannesburg", "Cape Town", "Durban", "Johannesburg"}

print(cities)

cities.add("Pretoria")
cities.discard("Durban")

print(cities)
```

### Comparing sets

Set operations can help compare groups of values:

- `intersection()` returns values present in both sets.
- `union()` returns all unique values from both sets.
- `difference()` returns values in the first set that are not in the second set.

The intersection is conceptually similar to finding matching values between two datasets, although it is not a full SQL join.

```python
department_a = {"Thabo", "John", "Sarah"}
department_b = {"Sarah", "Peter", "John"}

print("Common employees:", department_a.intersection(department_b))
print("All unique employees:", department_a.union(department_b))
print("Only in department A:", department_a.difference(department_b))
```

## 5. Combining data structures

Business datasets often contain multiple records. A common Python structure is a **list of dictionaries**:
- The outer list `[]` stores multiple records.
- Each inner dictionary `{}` stores the attributes of one record.

```python
customers = [
    {
        "customer_id": 1001,
        "name": "Thabo",
        "city": "Johannesburg",
        "total_spent": 15000.00
    },
    {
        "customer_id": 1002,
        "name": "Sarah",
        "city": "Cape Town",
        "total_spent": 22000.00
    },
    {
        "customer_id": 1003,
        "name": "John",
        "city": "Durban",
        "total_spent": 8500.00
    }
]

print(customers[0])
print(customers[0]["name"])
```

### Reading records with a loop

A `for` loop processes the customer records one at a time. In each repetition, `customer` refers to the current dictionary in the `customers` list.

This example uses a loop to print each customer's name.

```python
for customer in customers:
    print(customer["name"])
```

### Calculating total customer spending

Start with `total_spending = 0`. For each customer, add that customer's `total_spent` value to the running total using `+=`.

`+=` is a shorthand for adding a value to the current variable and assigning the result back to that variable.

```python
total_spending = 0

for customer in customers:
    total_spending += customer["total_spent"]

print("Total spending:", total_spending)
```

### Exercise 5 — Mini data engineering challenge

The transaction records below contain a transaction ID, product, quantity, and unit price.

For each record:
1. Calculate the individual transaction value using quantity × unit price.
2. Display the transaction ID, product, and calculated value.
3. Add the individual value to the overall sales total.
4. Convert the records into a Pandas DataFrame.
5. Add a `transaction_value` column to the DataFrame.

The individual transaction value and the overall total are separate calculations.

```python
transactions = [
    {
        "transaction_id": 1001,
        "product": "Laptop",
        "quantity": 2,
        "unit_price": 12500.00
    },
    {
        "transaction_id": 1002,
        "product": "Keyboard",
        "quantity": 5,
        "unit_price": 450.00
    },
    {
        "transaction_id": 1003,
        "product": "Monitor",
        "quantity": 3,
        "unit_price": 3200.00
    }
]

total_sales = 0

for record in transactions:
    transaction_value = record["quantity"] * record["unit_price"]

    print("Transaction ID:", record["transaction_id"])
    print("Product:", record["product"])
    print("Transaction value:", transaction_value)

    total_sales += transaction_value

print("Total sales:", total_sales)
```

```python
import pandas as pd

df = pd.DataFrame(transactions)
df["transaction_value"] = df["quantity"] * df["unit_price"]

print(df)
```
