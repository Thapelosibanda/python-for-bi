## 1. Variables and assignment

A variable is a name that refers to a value. The assignment operator `=` assigns the value on the right to the variable name on the left.

In the example below:
- `product_name` stores text.
- `quantity_sold` stores a whole number.
- `unit_price` stores a decimal number.

Variable names should be descriptive. Use lowercase words separated by underscores, as shown below.

```python
product_name = "Laptop"
quantity_sold = 15
unit_price = 12500.00

print(product_name)
print(quantity_sold)
print(unit_price)
```

## 2. Common Python data types

| Data type | Meaning | Example |
|---|---|---|
| `str` | Text/string | `"Laptop"` |
| `int` | Whole number | `15` |
| `float` | Number with a decimal component | `12500.50` |
| `bool` | Boolean value: `True` or `False` | `True` |

Use `type()` to inspect the data type of a value or variable.

```python
product_name = "Laptop"
quantity_sold = 15
unit_price = 12500.50
is_available = True

print(type(product_name))
print(type(quantity_sold))
print(type(unit_price))
print(type(is_available))
```

## 3. Arithmetic operators

Python supports common arithmetic operations:

| Operator | Operation |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `//` | Floor division: division rounded down to a whole-number result |
| `%` | Modulo: remainder after division |
| `**` | Exponentiation: power |

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

## 4. A practical sales calculation

A business sells 24 wireless mice at R349.99 each and offers a 10% discount.

The calculation is performed in three stages:

1. Gross revenue = quantity × unit price.
2. Discount amount = gross revenue × discount rate.
3. Net revenue = gross revenue − discount amount.

`discount_rate = 0.10` represents 10% as a decimal.

```python
product_name = "Wireless Mouse"
quantity_sold = 24
unit_price = 349.99
discount_rate = 0.10

gross_revenue = quantity_sold * unit_price
discount_amount = gross_revenue * discount_rate
net_revenue = gross_revenue - discount_amount

print("Product:", product_name)
print("Gross revenue:", gross_revenue)
print("Discount:", discount_amount)
print("Net revenue:", net_revenue)
```

## 5. Type conversion

Data received from files or external systems may represent numbers as text.

For example, `"25"` is a string, not an integer. Multiplying a string by `2` repeats the text rather than performing numeric multiplication.

```python
quantity = "25"

print(type(quantity))
print(quantity * 2)
```

Convert the string to an integer with `int()` when you need to perform numeric calculations. Use `float()` when the value contains a decimal component.

Not every value that looks numeric should be converted. An identifier such as `"000789"` should generally remain a string because converting it to an integer would remove its leading zeros.

```python
quantity = "25"
quantity = int(quantity)

print(quantity)
print(type(quantity))

unit_price = "45.75"
unit_price = float(unit_price)

customer_id = "000789"

total_transaction = quantity * unit_price

print("Unit price:", unit_price)
print("Customer ID:", customer_id)
print("Customer ID data type:", type(customer_id))
print("Total transaction value:", total_transaction)
```

## 6. Comparison operators

Comparison operators compare values and return a Boolean result: `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

Important: `=` assigns a value; `==` checks whether two values are equal.

```python
sales_target = 100000
actual_sales = 125000

print(actual_sales > sales_target)
print(actual_sales == sales_target)
print(actual_sales < sales_target)
```

### Exercise 3 — Data type conversion

The following values arrive as strings:

Convert quantity and unit price to numeric types, calculate the transaction value, and keep the customer ID as text so its leading zeros are preserved.
