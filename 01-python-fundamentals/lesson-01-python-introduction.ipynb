## 1. What is Python?

Python is a general-purpose programming language. In Business Intelligence and data engineering, it is commonly used to automate repetitive tasks, inspect and transform data, perform analysis, and support data pipelines.

A Jupyter notebook lets you write and run Python in individual cells. You can run a cell, inspect its output, and then continue with the next cell.

## 2. Your first Python instruction

`print()` displays information in the notebook output.

The text `"Hello, world!"` is a string: text enclosed in quotation marks.

```python
print("Hello, world!")
```

## 3. Running notebook cells

A notebook is made up of cells. Code cells contain Python instructions. Run a code cell to execute its instructions.

You can run cells independently, but a cell may depend on variables or imports created by an earlier cell. If you restart the kernel, run the required cells again from the top.

## 4. Inspecting the Python environment

The `sys` module provides information about the Python interpreter. The `pandas` and `numpy` libraries are widely used in data analysis.

- `import` makes a module available in the current notebook.
- `sys.version` reports the Python version.
- `sys.executable` shows the Python interpreter executable being used.
- `pd.__version__` and `np.__version__` show the installed library versions.

If a library is not installed in the active environment, its import will produce an error.

```python
import sys
import pandas as pd
import numpy as np

print("Python version:", sys.version)
print("Python executable:", sys.executable)
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

## 5. Why environment information matters

When working on a data project, the Python environment matters because different projects may use different Python versions or library versions.

Checking the interpreter and library versions helps you confirm which environment is running your code and makes troubleshooting easier.
